```python
import numpy as np
import matplotlib.pyplot as plt


# =====================================================================
# Code mẫu: huấn luyện đầy đủ và quan sát loss giảm
# =====================================================================

# Dữ liệu mẫu: y gần đúng bằng 3x + 2
# Dùng x trong khoảng 0 đến 1 để loss giảm mượt và dễ quan sát hơn.
x = np.linspace(0, 1, 20)
y = 3 * x + 2

# Khởi tạo trọng số
w = 0.0
b = 0.0
learning_rate = 0.1
epochs = 80
loss_history = []


def predict(x, w, b):
    return w * x + b


def mse_loss(y_true, y_pred):
    error = y_pred - y_true
    return np.mean(error ** 2)


def compute_gradients(x, y_true, y_pred):
    error = y_pred - y_true
    grad_w = 2 * np.mean(x * error)
    grad_b = 2 * np.mean(error)
    return grad_w, grad_b


for epoch in range(epochs):
    # Tính giá trị dự đoán
    y_pred = predict(x, w, b)

    # Tính loss
    loss = mse_loss(y, y_pred)
    loss_history.append(loss)

    # Tính gradient
    grad_w, grad_b = compute_gradients(x, y, y_pred)

    # Cập nhật trọng số
    w = w - learning_rate * grad_w
    b = b - learning_rate * grad_b

    if epoch % 20 == 0 or epoch == epochs - 1:
        print(f"epoch {epoch:3d} | loss = {loss:8.4f} | w = {w:6.3f} | b = {b:6.3f}")


def plot_training_result(loss_history, x, y_true, w, b):
    epochs_axis = np.arange(len(loss_history))
    y_pred = predict(x, w, b)

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    axes[0].plot(epochs_axis, loss_history, color="tab:blue", linewidth=2)
    axes[0].set_title("Loss theo epoch")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss MSE")
    axes[0].grid(True, alpha=0.3)

    axes[1].scatter(x, y_true, color="tab:orange", label="Dữ liệu thật")
    axes[1].plot(x, y_pred, color="tab:blue", linewidth=2, label="Đường dự đoán")
    axes[1].set_title(f"Đường thẳng học được: y = {w:.2f}x + {b:.2f}")
    axes[1].set_xlabel("x")
    axes[1].set_ylabel("y")
    axes[1].grid(True, alpha=0.3)
    axes[1].legend()

    plt.tight_layout()
    plt.savefig("sample_training_result.png", dpi=100)
    plt.close(fig)


plot_training_result(loss_history, x, y, w, b)

print("\nKết quả cuối cùng:")
print(f"w = {w:.4f}")
print(f"b = {b:.4f}")
print(f"Dự đoán khi x = 1.5: {predict(1.5, w, b):.4f}")

# =====================================================================
# Bài 1: Linear Regression với dữ liệu rải quanh một đường thẳng
# =====================================================================
#
# Bộ dữ liệu bên dưới mô phỏng các điểm x nằm rải rác quanh đường thẳng
# thật gần đúng y = 2.4x - 1.2.
#
# Yêu cầu:
# - Tính dự đoán y_pred.
# - Tính MSE loss.
# - Tính gradient theo w và b.
# - Cập nhật w, b bằng Gradient Descent.
# - In loss để quan sát quá trình học.

# Dữ liệu linear regression
x_linear = np.array([-3.0, -2.4, -1.8, -1.2, -0.6, 0.0, 0.6, 1.2, 1.8, 2.4, 3.0])
y_linear = np.array([-8.7, -6.6, -5.4, -4.0, -2.4, -1.0, 0.0, 1.6, 3.0, 4.7, 6.2])

# Tham số khởi tạo đã cho
w_linear = 0.3
b_linear = 0.0
learning_rate_linear = 0.03
epochs_linear = 120

print("\n" + "=" * 60)
print("BÀI 1: LINEAR REGRESSION")
print("=" * 60)
print("x_linear =", x_linear)
print("y_linear =", y_linear)
print("w_linear ban đầu =", w_linear)
print("b_linear ban đầu =", b_linear)
print("learning_rate_linear =", learning_rate_linear)


# --- Bài 1.1: Hoàn thiện hàm cho Linear Regression ---

def predict_linear(x, w, b):
    """Hàm dự đoán y_pred = w * x + b."""
    return w * x + b


# mse_loss

def compute_linear_gradients(x, y_true, y_pred):
    """Tính gradient của MSE loss theo w và b."""
    error = y_pred - y_true
    grad_w = 2 * np.mean(x * error)
    grad_b = 2 * np.mean(error)
    return grad_w, grad_b
```

```python
# --- Bài 1.2: Huấn luyện mô hình Linear Regression ---

loss_history_linear = []

for epoch in range(epochs_linear):
    # Tính y_pred_linear bằng predict_linear
    y_pred_linear = predict_linear(x_linear, w_linear, b_linear)

    # Tính loss_linear bằng mse_loss và lưu vào loss_history_linear
    loss_linear = mse_loss(y_linear, y_pred_linear)
    loss_history_linear.append(loss_linear)

    # Tính grad_w_linear, grad_b_linear bằng compute_linear_gradients
    grad_w_linear, grad_b_linear = compute_linear_gradients(x_linear, y_linear, y_pred_linear)

    # Cập nhật w_linear, b_linear
    w_linear = w_linear - learning_rate_linear * grad_w_linear
    b_linear = b_linear - learning_rate_linear * grad_b_linear

    # In epoch, loss_linear, w_linear, b_linear sau mỗi 20 epochs
    if epoch % 20 == 0 or epoch == epochs_linear - 1:
        print(f"epoch {epoch:3d} | loss = {loss_linear:8.4f} | "
              f"w = {w_linear:6.3f} | b = {b_linear:6.3f}")

# In kết quả cuối cùng của w_linear và b_linear
print("\nKết quả cuối cùng (Linear Regression):")
print(f"w_linear = {w_linear:.4f}")
print(f"b_linear = {b_linear:.4f}")
```

```python
# --- Bài 1.3: Viết hàm plot kết quả Linear Regression ---
import matplotlib.pyplot as plt
def plot_linear_result(x, y_true, w, b, loss_history):
    """Vẽ 2 biểu đồ:
    - Biểu đồ 1: loss theo epoch
    - Biểu đồ 2: scatter dữ liệu thật + đường y_pred = w*x + b
    """
    epochs_axis = np.arange(len(loss_history))
    y_pred = predict_linear(x, w, b)

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    axes[0].plot(epochs_axis, loss_history, color="tab:blue", linewidth=2)
    axes[0].set_title("Bài 1: Loss theo epoch")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss MSE")
    axes[0].grid(True, alpha=0.3)

    axes[1].scatter(x, y_true, color="tab:orange", label="Dữ liệu thật")
    axes[1].plot(x, y_pred, color="tab:blue", linewidth=2, label="Đường dự đoán")
    axes[1].set_title(f"Bài 1: y = {w:.2f}x + {b:.2f}")
    axes[1].set_xlabel("x")
    axes[1].set_ylabel("y")
    axes[1].grid(True, alpha=0.3)
    axes[1].legend()

    plt.tight_layout()
    plt.savefig("bai1_linear_result.png", dpi=100)
    plt.close(fig)


plot_linear_result(x_linear, y_linear, w_linear, b_linear, loss_history_linear)
```

```python
# =====================================================================
# Bài 2: Logistic Regression
# =====================================================================
#
# Bộ dữ liệu bên dưới mô phỏng một biến đầu vào x. Nhãn y = 0 thường nằm
# ở vùng x nhỏ hơn, nhãn y = 1 thường nằm ở vùng x lớn hơn.
#
# Yêu cầu:
# - Tính dự đoán z từ trọng số.
# - Đổi z thành xác suất bằng sigmoid.
# - Tính Binary Cross Entropy loss.
# - Tính gradient theo w và b.
# - Cập nhật w, b bằng Gradient Descent.

# Dữ liệu logistic regression
x_logistic = np.array([-3.2, -2.7, -2.1, -1.5, -0.9, -0.4, 0.1, 0.5, 0.9, 1.4, 2.0, 2.6, 3.1])
y_logistic = np.array([0, 0, 0, 0, 0, 1, 0, 1, 1, 1, 1, 1, 1])

# Tham số khởi tạo
w_logistic = -0.2
b_logistic = 0.1
learning_rate_logistic = 0.15
epochs_logistic = 150

print("\n" + "=" * 60)
print("BÀI 2: LOGISTIC REGRESSION")
print("=" * 60)
print("x_logistic =", x_logistic)
print("y_logistic =", y_logistic)
print("w_logistic ban đầu =", w_logistic)
print("b_logistic ban đầu =", b_logistic)
print("learning_rate_logistic =", learning_rate_logistic)
```

```python
# =====================================================================
# Bài 2: Logistic Regression
# =====================================================================
#
# Bộ dữ liệu bên dưới mô phỏng một biến đầu vào x. Nhãn y = 0 thường nằm
# ở vùng x nhỏ hơn, nhãn y = 1 thường nằm ở vùng x lớn hơn.
#
# Yêu cầu:
# - Tính dự đoán z từ trọng số.
# - Đổi z thành xác suất bằng sigmoid.
# - Tính Binary Cross Entropy loss.
# - Tính gradient theo w và b.
# - Cập nhật w, b bằng Gradient Descent.

# Dữ liệu logistic regression
x_logistic = np.array([-3.2, -2.7, -2.1, -1.5, -0.9, -0.4, 0.1, 0.5, 0.9, 1.4, 2.0, 2.6, 3.1])
y_logistic = np.array([0, 0, 0, 0, 0, 1, 0, 1, 1, 1, 1, 1, 1])

# Tham số khởi tạo
w_logistic = -0.2
b_logistic = 0.1
learning_rate_logistic = 0.15
epochs_logistic = 150

print("\n" + "=" * 60)
print("BÀI 2: LOGISTIC REGRESSION")
print("=" * 60)
print("x_logistic =", x_logistic)
print("y_logistic =", y_logistic)
print("w_logistic ban đầu =", w_logistic)
print("b_logistic ban đầu =", b_logistic)
print("learning_rate_logistic =", learning_rate_logistic)


# --- Bài 2.1: Hoàn thiện hàm cho Logistic Regression ---

def sigmoid(z):
    """Hàm sigmoid: chuyển z thành giá trị trong khoảng (0, 1)."""
    return 1 / (1 + np.exp(-z))


def predict_proba_logistic(x, w, b):
    """Tính z = w*x + b rồi đổi thành xác suất bằng sigmoid."""
    z = w * x + b
    return sigmoid(z)


def binary_cross_entropy(y_true, y_pred_proba):
    """Tính Binary Cross Entropy loss trung bình cho n mẫu.
    Dùng epsilon để tránh log(0).
    """
    epsilon = 1e-12
    y_pred_proba = np.clip(y_pred_proba, epsilon, 1 - epsilon)
    loss = -np.mean(
        y_true * np.log(y_pred_proba) + (1 - y_true) * np.log(1 - y_pred_proba)
    )
    return loss


def compute_logistic_gradients(x, y_true, y_pred_proba):
    """Gradient gọn nhờ kết hợp sigmoid + BCE:
    dL/dw = (1/n) * sum(x_i * (y_hat_i - y_i))
    dL/db = (1/n) * sum(y_hat_i - y_i)
    """
    error = y_pred_proba - y_true
    grad_w = np.mean(x * error)
    grad_b = np.mean(error)
    return grad_w, grad_b


# --- Bài 2.2: Huấn luyện mô hình Logistic Regression ---

loss_history_logistic = []

for epoch in range(epochs_logistic):
    # Tính y_pred_proba_logistic bằng predict_proba_logistic
    y_pred_proba_logistic = predict_proba_logistic(x_logistic, w_logistic, b_logistic)

    # Tính loss_logistic bằng binary_cross_entropy và lưu vào loss_history_logistic
    loss_logistic = binary_cross_entropy(y_logistic, y_pred_proba_logistic)
    loss_history_logistic.append(loss_logistic)

    # Tính grad_w_logistic, grad_b_logistic bằng compute_logistic_gradients
    grad_w_logistic, grad_b_logistic = compute_logistic_gradients(
        x_logistic, y_logistic, y_pred_proba_logistic
    )

    # Cập nhật w_logistic, b_logistic
    w_logistic = w_logistic - learning_rate_logistic * grad_w_logistic
    b_logistic = b_logistic - learning_rate_logistic * grad_b_logistic

    # In epoch, loss_logistic, w_logistic, b_logistic sau mỗi 25 epochs
    if epoch % 25 == 0 or epoch == epochs_logistic - 1:
        print(f"epoch {epoch:3d} | loss = {loss_logistic:8.4f} | "
              f"w = {w_logistic:6.3f} | b = {b_logistic:6.3f}")

# Tính xác suất cuối cùng cho từng điểm x_logistic
final_proba_logistic = predict_proba_logistic(x_logistic, w_logistic, b_logistic)
# Đổi xác suất thành nhãn dự đoán với ngưỡng 0.5
predicted_labels_logistic = (final_proba_logistic >= 0.5).astype(int)

# In xác suất, nhãn dự đoán, và so sánh với y_logistic
print("\nKết quả cuối cùng (Logistic Regression):")
print(f"w_logistic = {w_logistic:.4f}")
print(f"b_logistic = {b_logistic:.4f}")
print(f"{'x':>6} | {'y_true':>6} | {'proba':>7} | {'y_pred':>6} | {'đúng?':>6}")
for xi, yi, pi, li in zip(x_logistic, y_logistic, final_proba_logistic, predicted_labels_logistic):
    dung = "Đúng" if yi == li else "Sai"
    print(f"{xi:6.2f} | {yi:6d} | {pi:7.4f} | {li:6d} | {dung:>6}")

so_dung = np.sum(predicted_labels_logistic == y_logistic)
print(f"\nSố dự đoán đúng: {so_dung}/{len(y_logistic)}")


# --- Bài 2.3: Viết hàm plot kết quả Logistic Regression ---

def plot_logistic_result(x, y_true, y_pred_proba, loss_history):
    """Vẽ 2 biểu đồ:
    - Biểu đồ 1: loss theo epoch
    - Biểu đồ 2: scatter dữ liệu thật + scatter xác suất dự đoán,
      kèm đường ngang y = 0.5 làm ngưỡng phân lớp.
    """
    epochs_axis = np.arange(len(loss_history))

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    axes[0].plot(epochs_axis, loss_history, color="tab:blue", linewidth=2)
    axes[0].set_title("Bài 2: Loss theo epoch")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss BCE")
    axes[0].grid(True, alpha=0.3)

    axes[1].scatter(x, y_true, color="tab:orange", label="Nhãn thật (0/1)")
    axes[1].scatter(x, y_pred_proba, color="tab:blue", marker="x", label="Xác suất dự đoán")
    axes[1].axhline(0.5, color="gray", linestyle="--", linewidth=1, label="Ngưỡng 0.5")
    axes[1].set_title("Bài 2: Xác suất dự đoán vs nhãn thật")
    axes[1].set_xlabel("x")
    axes[1].set_ylabel("y / xác suất")
    axes[1].grid(True, alpha=0.3)
    axes[1].legend()

    plt.tight_layout()
    plt.savefig("bai2_logistic_result.png", dpi=100)
    plt.close(fig)


plot_logistic_result(x_logistic, y_logistic, final_proba_logistic, loss_history_logistic)

print("\nĐã lưu các biểu đồ: sample_training_result.png, "
      "bai1_linear_result.png, bai2_logistic_result.png")
```
